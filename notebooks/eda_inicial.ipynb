Importação e visualização inicial do extrato bancário

```python
import pandas as pd
from pathlib import Path

BASE_DIR = Path.cwd().parent
DATA_DIR = BASE_DIR / "data"
RAW_FILE = DATA_DIR / "raw/Bradesco_29042026_080947.CSV"

df = pd.read_csv(
    RAW_FILE,
    sep=';',
    skiprows=2,
    encoding='latin1'
)

df
```

Vou fazer uma pré limpeza, só pra exibição e testar os dados

```python
colunas = ['data', 'lancamento', 'dcto', 'credito', 'debito', 'saldo']

df.columns = colunas
df
```

```python
df['lancamento'] = (
    df['lancamento']
    .str.replace(r'^TRANSFERENCIA PIX REM:\s*', '', regex=True)
    .str.replace(r'\s+\d{2}/\d{2}$', '', regex=True)
    .str.strip()
)

df
```
